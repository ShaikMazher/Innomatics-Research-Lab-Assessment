```python
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate

print("="*50)
print("TASK 1: Replace Hardcoded Prompts")
print("="*50)
# Requirements: No f-strings, use PromptTemplate, must be reusable[cite: 1]
task1_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners"
)
print(task1_template.format(topic="Generative AI"))


print("\n" + "="*50)
print("TASK 2: Multi-Input Prompt System")
print("="*50)
# Build a template with topic, audience, and tone[cite: 1]
task2_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

test_cases = [
    {"topic": "AI", "audience": "beginners", "tone": "friendly"},
    {"topic": "Python", "audience": "kids", "tone": "fun"},
    {"topic": "Deep Learning", "audience": "engineers", "tone": "technical"}
]

for case in test_cases:
    print(f"- {task2_template.format(**case)}")


print("\n" + "="*50)
print("TASK 3: Prompt Variations Engine")
print("="*50)
# Create 3 distinct templates and run for "Machine Learning"[cite: 1]
teaching_template = PromptTemplate(input_variables=["topic"], template="Explain {topic} clearly step by step")
interview_template = PromptTemplate(input_variables=["topic"], template="Ask 3 questions about {topic}")
story_template = PromptTemplate(input_variables=["topic"], template="Explain {topic} as a story")

target_topic = "Machine Learning"
print("Teaching:  ", teaching_template.format(topic=target_topic))
print("Interview: ", interview_template.format(topic=target_topic))
print("Story:     ", story_template.format(topic=target_topic))


print("\n" + "="*50)
print("TASK 4: ChatPrompt Template System")
print("="*50)
# Build a chat-based system using roles[cite: 1]
chat_template = ChatPromptTemplate.from_messages([
    ("system", "You are a {role}. Act accordingly and guide the user."),
    ("user", "The topic is {topic}.")
])

roles = ["teacher", "interviewer", "motivator"]
for role in roles:
    formatted_chat = chat_template.format_messages(role=role, topic="Neural Networks")
    print(f"\n--- Role: {role.upper()} ---")
    for msg in formatted_chat:
        print(f"{msg.type.capitalize()}: {msg.content}")


print("\n" + "="*50)
print("TASK 5: Input Validation Layer")
print("="*50)
# Validate audience and tone, assign default values if invalid[cite: 1]
def validate_inputs(audience, tone):
    valid_audiences = ["beginner", "intermediate", "expert"]
    valid_tones = ["formal", "casual", "fun"]

    final_audience = audience if audience in valid_audiences else "beginner"
    final_tone = tone if tone in valid_tones else "casual"

    if final_audience != audience or final_tone != tone:
        print(f"[Warning] Invalid inputs detected. Using defaults -> Audience: {final_audience}, Tone: {final_tone}")

    return final_audience, final_tone

# Testing the validation
print("Valid input test:", validate_inputs("expert", "formal"))
print("Invalid input test:", validate_inputs("super_genius", "angry"))


print("\n" + "="*50)
print("TASK 6: Prompt Generator App")
print("="*50)
# Combine validation and prompt generation[cite: 1]
def generate_prompt(topic, audience, tone, style):
    # Pass through Task 5 validation
    valid_audience, valid_tone = validate_inputs(audience, tone)

    app_template = PromptTemplate(
        input_variables=["topic", "audience", "tone", "style"],
        template="Explain {topic} for {audience} in a {tone} {style} style"
    )
    return app_template.format(topic=topic, audience=valid_audience, tone=valid_tone, style=style)

print(generate_prompt("Neural Networks", "beginner", "fun", "storytelling"))


print("\n" + "="*50)
print("TASK 7: Template Reusability Test")
print("="*50)
# Run one template with 5 different inputs to prove dynamic reusability[cite: 1]
reusable_template = PromptTemplate(
    input_variables=["action", "subject"],
    template="Please {action} the {subject}."
)

reusability_tests = [
    {"action": "summarize", "subject": "article"},
    {"action": "translate", "subject": "document into Spanish"},
    {"action": "analyze", "subject": "sales data"},
    {"action": "debug", "subject": "Python script"},
    {"action": "rewrite", "subject": "email professionally"}
]

for i, test in enumerate(reusability_tests, 1):
    print(f"Run {i}: {reusable_template.format(**test)}")
```
