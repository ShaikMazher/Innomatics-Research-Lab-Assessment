```python
# Problem 1: Smart Parking Lot
def manage_parking(capacity, logs):
    parked_vehicles = 0

    for action in logs:
        if action == "IN":
            parked_vehicles += 1
        elif action == "OUT":
            parked_vehicles -= 1

    print(f"Currently Parked Vehicles: {parked_vehicles}")

    if parked_vehicles >= capacity:
        print("Parking Status: Capacity Exceeded / Full")
    else:
        print("Parking Status: Available")

# Test Data
manage_parking(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

```python
# Problem 2: Food Delivery ETA
def estimate_delivery(distance, traffic, weather):
    # Base time: roughly 5 minutes per km
    estimated_time = distance * 5

    if traffic.lower() == "high":
        estimated_time += 10

    if weather.lower() == "rainy":
        estimated_time += 5

    print(f"Estimated Delivery Time: {estimated_time} minutes")

# Test Data
estimate_delivery(8, "High", "Rainy")
```

```python
# Problem 3: Theatre Occupancy
def analyze_occupancy(total_seats, booked_seats_list):
    booked_count = len(booked_seats_list)
    occupancy_percent = (booked_count / total_seats) * 100

    print(f"Occupancy: {int(occupancy_percent)}%")

    if occupancy_percent == 100:
        print("Show Status: Housefull")
    elif occupancy_percent >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Available")

# Test Data (creating a dummy list of 150 booked seats)
booked = [1] * 150
analyze_occupancy(200, booked)
```

```python
# Problem 4: CPU Load Analyzer
def classify_server_load(cpu_readings):
    total = sum(cpu_readings)
    average = total / len(cpu_readings)

    print(f"Average CPU Load: {int(average)}%")

    if average < 50:
        print("Server Status: Normal")
    elif average <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")

# Test Data
classify_server_load([45, 60, 70, 85, 90])
```

```python
# Problem 5: Resource Overuse (Assuming > 8 hours is overuse)
def monitor_resources(usage_data):
    overused = []

    for resource, hours in usage_data.items():
        if hours > 8:
            overused.append(resource)

    if overused:
        # Convert list to string for clean printing
        print(f"Overused Resources: {', '.join(overused)}")
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")

# Test Data
monitor_resources({"Projector": 6, "AC": 9, "Lights": 4})
```

```python
# Problem 6: Event Registration
def manage_event(capacity, registrations):
    if registrations <= capacity:
        confirmed = registrations
        waitlisted = 0
        status = "Open"
    else:
        confirmed = capacity
        waitlisted = registrations - capacity
        status = "Closed"

    print(f"Confirmed Registrations: {confirmed}")
    print(f"Waitlisted Users: {waitlisted}")
    print(f"Registration Status: {status}")

# Test Data
manage_event(100, 105)
```
