```python
# Task 2 - Python Practice Problems
# completed by Shaik Mazher

# 1. Unique Words
sentence = "python is easy and python is powerful"
words_list = sentence.split()
unique_words = set(words_list)

print("Unique words count:", len(unique_words))
print("Unique words:", unique_words)
print("-------------------")

# 2. Highest Salary
employees = { "Ravi": 75000, "Anita": 68000, "Kiran": 72000 }
max_sal = 0
best_emp = ""

# finding the max salary with a loop
for name in employees:
    if employees[name] > max_sal:
        max_sal = employees[name]
        best_emp = name

print("Highest Salary:", best_emp, "-", max_sal)
```

```python
# 3. Max and Min values
numbers = [45, 22, 89, 10, 66]
max_val = numbers[0]
min_val = numbers[0]

for num in numbers:
    if num > max_val:
        max_val = num
    if num < min_val:
        min_val = num

print("List:", numbers)
print("Max:", max_val)
print("Min:", min_val)
print("-------------------")

# 4. Count Products Above 1000
prices = [450, 1200, 899, 1500, 300]
count = 0

for p in prices:
    if p > 1000:
        count = count + 1

print("Products above 1000:", count)
print("-------------------")

# 5. Attendance Percentage
attendance = ["P", "P", "A", "P", "P"]
p_count = 0

for day in attendance:
    if day == "P":
        p_count += 1

percent = (p_count / len(attendance)) * 100
print("Attendance Percentage:", percent)
print("-------------------")

# 6. Remove Duplicate Phones
phone_numbers = [9876543210, 9123456789, 9876543210]
unique_phones = set(phone_numbers)

print("Unique phone numbers:", unique_phones)

```

```python
# 7. Character Frequency
word = "pythonp"
char_count = {}

# counting letters
for c in word:
    if c in char_count:
        char_count[c] = char_count[c] + 1
    else:
        char_count[c] = 1

print(char_count)
print("-------------------")

# 8. Convert List to Tuple
my_list = [10, 20, 30]
my_tuple = tuple(my_list)

print("Tuple:", my_tuple)
print("-------------------")

# 9. Check key in dict
student_data = {"name": "Mazher", "role": "intern"}

if "name" in student_data:
    print("Employee exists")
print("-------------------")

# 10. Average Marks
marks = [80, 65, 75, 70]
total = 0

for m in marks:
    total = total + m

avg = total / len(marks)
print("Average Marks:", avg)
```
