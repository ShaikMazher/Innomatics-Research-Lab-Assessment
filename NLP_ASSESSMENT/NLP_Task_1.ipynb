```python
import re

def preprocess_text(text):
    # Task 7: Error Handling - Return empty string if input is blank or invalid
    if not text or str(text).strip() == "":
        return ""

    # Requirement 4: Convert text to lowercase first to make everything uniform
    text = str(text).lower()

    # Requirement 6: Remove URLs and email-like patterns
    # 'http\S+' looks for http followed by any non-space characters
    text = re.sub(r'http\S+|www\.\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)

    # Requirement 1: Remove numbers[cite: 1]
    # \d+ matches any digits from 0-9
    text = re.sub(r'\d+', '', text)

    # Requirement 3: Handle repeated characters (e.g., sooo goooood -> soo good)[cite: 1]
    # This finds any character repeating 3 or more times and reduces it.
    # We replace it with '\1\1' (two characters) to prevent words like "good" from becoming "god".
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)

    # Extra cleaning: Remove punctuation so we don't get messy tokens like "course!!!"
    text = re.sub(r'[^\w\s]', ' ', text)

    # Requirement 5: Remove very short tokens (length <= 2) EXCEPT "no" and "not"[cite: 1]
    # Requirement 2: Splitting the text automatically handles removing extra spaces[cite: 1]
    words = text.split()
    clean_words = []

    for word in words:
        if len(word) > 2 or word in ["no", "not"]:
            clean_words.append(word)

    # Join the cleaned words back into a single sentence
    return " ".join(clean_words)
```

```python
# Task 3: Stress Testing
sample_inputs = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

print("--- Task 3: Stress Testing Results ---\n")

# Loop through each sentence to process and display the results
for text in sample_inputs:
    clean_sentence = preprocess_text(text)

    # Tokens are just the individual words separated into a list
    clean_tokens = clean_sentence.split()

    # Formatting the output exactly as requested in the assignment
    print(f"Original Text:    {text}")
    print(f"Cleaned Tokens:   {clean_tokens}")
    print(f"Cleaned Sentence: {clean_sentence}")
    print("-" * 60)
```

```python
# Task 4: Token Analytics

print("--- Task 4: Token Analytics Results ---\n")

for text in sample_inputs:
    clean_sentence = preprocess_text(text)
    tokens = clean_sentence.split()

    # 1. Total number of tokens[cite: 1]
    total_tokens = len(tokens)

    # 2. Number of unique tokens (using set() to remove duplicates)[cite: 1]
    unique_tokens = len(set(tokens))

    # 3. Average token length[cite: 1]
    if total_tokens > 0:
        avg_length = sum(len(word) for word in tokens) / total_tokens
    else:
        avg_length = 0.0

    print(f"Original: {text}")
    print(f"Total Tokens: {total_tokens} | Unique Tokens: {unique_tokens} | Avg Token Length: {avg_length:.2f}")
    print("-" * 60)
```

```python
# Task 5: Frequency Analysis[cite: 1]
from collections import Counter

# 1. Combine all tokens from all sentences into one massive list
all_tokens = []
for text in sample_inputs:
    clean_sentence = preprocess_text(text)
    # .extend() adds the individual words to our master list, rather than creating lists within lists
    all_tokens.extend(clean_sentence.split())

# 2. Use Counter to automatically tally up every word[cite: 1]
word_counts = Counter(all_tokens)

print("--- Task 5: Frequency Analysis ---")

# 3. Identify Top 10 most frequent words[cite: 1]
print("\nTop 10 Most Frequent Words:")
top_10 = word_counts.most_common(10)
for word, count in top_10:
    print(f"- {word}: {count}")

# 4. Identify Top 5 least frequent words[cite: 1]
# By leaving most_common() empty, it sorts everything. Then we grab the last 5 items using [-5:]
print("\nTop 5 Least Frequent Words:")
bottom_5 = word_counts.most_common()[-5:]
for word, count in bottom_5:
    print(f"- {word}: {count}")
```

```python
# ==========================================
# Task 6: Build Full Pipeline
# ==========================================
def full_pipeline(text_list):
    all_tokens = []
    all_clean_sentences = []

    for text in text_list:
        clean_sentence = preprocess_text(text)

        # We append the results to our lists
        all_clean_sentences.append(clean_sentence)

        # The prompt asks for a list of tokens, so we split the sentence
        all_tokens.append(clean_sentence.split())

    # Expected Output format returning a dictionary
    return {
        "tokens": all_tokens,
        "clean_sentences": all_clean_sentences
    }

# Let's test the pipeline
pipeline_results = full_pipeline(sample_inputs)
print("--- Task 6: Full Pipeline Results ---")
print(f"Total Sentences Processed: {len(pipeline_results['clean_sentences'])}")
print("Pipeline successfully returned a dictionary with keys:", pipeline_results.keys())
print("\n")


# ==========================================
# Task 7: Error Handling
# ==========================================
error_test_cases = [
    "",              # Empty string[cite: 1]
    "😊🚀🔥",          # Only emojis[cite: 1]
    "123 456 789"    # Only numbers[cite: 1]
]

print("--- Task 7: Error Handling Results ---")
for text in error_test_cases:
    clean_sentence = preprocess_text(text)
    print(f"Original Text: '{text}'")
    print(f"Cleaned Output: '{clean_sentence}'")
    print("-" * 40)
```
