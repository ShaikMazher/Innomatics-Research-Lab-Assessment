```python
# Problem 1: Validate mobile recharge with a while loop
def validate_recharge():
    valid_plans = [199, 299, 399, 599]

    while True:
        amount_str = input("Enter recharge amount (or type 'exit' to stop): ")

        if amount_str.lower() == 'exit':
            print("Exiting system.")
            break

        amount = int(amount_str)

        if amount >= 50 and amount in valid_plans:
            print("Recharge successful!")
            break
        else:
            print("Invalid plan. Amount must be >= 50 and a valid plan (199, 299, 399, 599). Try again.")

# Uncomment the line below to test the function in Colab
# validate_recharge()
```

```python
# Problem 2: Check inventory using a dictionary and loop
def check_inventory(stock_data):
    for product, quantity in stock_data.items():
        if quantity < 15:
            print(f"{product}: Reorder Alert")
        else:
            print(f"{product}: Stock OK")

# Testing the function
warehouse_stock = {"Laptops": 10, "Mice": 45, "Keyboards": 12, "Monitors": 20}
check_inventory(warehouse_stock)
```

```python
# Problem 3: Calculate average marks and determine Pass/Fail
def process_result(marks_list):
    total = 0
    for mark in marks_list:
        total += mark

    average = total / len(marks_list)

    if average >= 50:
        print(f"Average: {average} - Pass")
        return "Pass"
    else:
        print(f"Average: {average} - Fail")
        return "Fail"

# Testing the function
student_marks = [45, 60, 55, 70]
process_result(student_marks)
```

```python
# Problem 4: Cab fare calculator with peak hour logic and retry
def cab_fare_estimator():
    base_fare = 50

    while True:
        km = float(input("Enter distance traveled in km: "))
        is_peak = input("Is it peak hour? (yes/no): ").strip().lower()

        fare = base_fare + (12 * km)

        if is_peak == 'yes':
            fare = fare + (fare * 0.25)  # Add 25% extra

        print(f"Estimated Cab Fare: ₹{fare}")

        retry = input("Do you want to calculate another fare? (yes/no): ").strip().lower()
        if retry != 'yes':
            print("Thank you for using the estimator!")
            break

# Uncomment the line below to test the function in Colab
# cab_fare_estimator()
```

```python
# Problem 5: Calculate attendance percentage
def check_eligibility(attendance_record):
    present_days = 0

    for status in attendance_record:
        if status == "P":
            present_days += 1

    percentage = (present_days / len(attendance_record)) * 100

    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"

# Testing the function
intern_attendance = ["P", "P", "A", "P", "P", "A", "P"]
result = check_eligibility(intern_attendance)
print(f"Attendance Status: {result}")
```

```python
# Problem 6: Validate password strength using string loops
def check_password_strength(password):
    if len(password) < 8:
        return "Weak: Password must be at least 8 characters long."

    has_digit = False
    has_special = False

    for char in password:
        if char.isdigit():
            has_digit = True
        if char in ['@', '#', '$']:
            has_special = True

    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak: Must contain at least one digit and one special character (@, #, $)."

# Testing the function
print(check_password_strength("Mazher123"))
print(check_password_strength("Mazher@123"))
```
