```python
# Problem 1: Smart Light Controller using OOP
class SmartLight:
    def __init__(self, name):
        self.name = name
        self.status = "OFF"

    def turn_on(self):
        self.status = "ON"
        print(f"{self.name} is {self.status}") # Expected Output: Bedroom Light is ON

    def turn_off(self):
        self.status = "OFF"
        print(f"{self.name} is {self.status}")

# Test Data
light = SmartLight("Bedroom Light")
light.turn_on()
```

```python
# Problem 2: Employee ID Card System
class EmployeeIDCard:
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department

    def display_card(self):
        # Displaying formatted ID card details[cite: 6]
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.department}")

# Test Data[cite: 6]
emp = EmployeeIDCard("Rahul", "EMP102", "IT")
emp.display_card()
```

```python
# Problem 3: Mobile Contact Record[cite: 6]
class MobileContact:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    def display_contact(self):
        # Displaying contact information[cite: 6]
        print("Contact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")

# Test Data[cite: 6]
contact = MobileContact("Anita", "9876543210")
contact.display_contact()
```

```python
# Problem 4: Product Price Tag Generator[cite: 6]
class ProductPriceTag:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display_tag(self):
        # Printing a formatted price tag with ₹ symbol[cite: 6]
        print(f"Product: {self.name}")
        print(f"Price: ₹{self.price}")

# Test Data[cite: 6]
product = ProductPriceTag("Headphones", 2499)
product.display_tag()
```

```python
# Problem 5: Movie Rating Display System[cite: 6]
class MovieRating:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    def display_rating(self):
        # Displaying movie details out of 5[cite: 6]
        print(f"Movie: {self.name}")
        print(f"Rating: {self.rating}/5")

# Test Data[cite: 6]
movie = MovieRating("Inception", 4.8)
movie.display_rating()
```

```python
# Problem 6: Delivery Address Manager[cite: 6]
class DeliveryManager:
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address

    def display_delivery(self):
        # Printing delivery details[cite: 6]
        print("Delivery Details")
        print(f"Customer: {self.customer_name}")
        print(f"Address: {self.address}")

# Test Data[cite: 6]
delivery = DeliveryManager("Suman", "Hyderabad")
delivery.display_delivery()
```
