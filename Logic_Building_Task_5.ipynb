```python
# Problem 1: Analyze post engagement
def analyze_engagement(likes):
    print(f"Total Likes: {likes}")
    if likes >= 1000:
        print("Post Status: Viral Post\n")
    else:
        print("Post Status: Normal Engagement\n")

# Test the function
analyze_engagement(1050)
```

```python
# Problem 2: Monitor medicine stock levels
def check_medicine_stock(stock):
    print(f"Medicine Stock: {stock}")
    if stock < 10:
        print("Status: Low Stock Alert\n")
    else:
        print("Status: Stock Sufficient\n")

# Test the function
check_medicine_stock(6)
```

```python
# Problem 3: Analyze seasonal rainfall data
def check_rainfall(rainfall_data, required_level):
    total = 0
    for rain in rainfall_data:
        total += rain

    avg_rainfall = total / len(rainfall_data)

    print(f"Average Rainfall: {avg_rainfall}")
    if avg_rainfall >= required_level:
        print("Rainfall Status: Adequate Rainfall\n")
    else:
        print("Rainfall Status: Inadequate Rainfall\n")

# Test the function
check_rainfall([70, 75, 71], 70)
```

```python
# Problem 4: Detect duplicate usernames using a Set
def detect_duplicate_accounts(usernames):
    unique_users = set(usernames)

    # If the set length is smaller than the list length, there are duplicates
    if len(unique_users) < len(usernames):
        print("Duplicate Accounts Found: Yes\n")
    else:
        print("Duplicate Accounts Found: No\n")

# Test the function
detect_duplicate_accounts(["user1", "user2", "user1", "user3"])
```

```python
# Problem 5: Check appointment eligibility based on age
def check_appointment(age):
    print(f"Patient Age: {age}")
    if age >= 18:
        print("Eligibility Status: Eligible\n")
    else:
        print("Eligibility Status: Not Eligible\n")

# Test the function
check_appointment(21)
```

```python
# Problem 6: Filter crop prices above 2000
def filter_premium_crops(prices):
    premium_crops = []

    for price in prices:
        if price > 2000:
            premium_crops.append(price)

    print(f"Premium Crops: {premium_crops}\n")

# Test the function
filter_premium_crops([1500, 2500, 1800, 3200])
```

```python
# Problem 7: Evaluate health based on error counts
def check_app_health(errors):
    print(f"Error Count: {errors}")
    if errors == 0:
        print("System Status: Healthy\n")
    elif errors <= 5:
        print("System Status: Minor Issues\n")
    else:
        print("System Status: Critical Issues\n")

# Test the function
check_app_health(7)
```

```python
# Problem 8: Validate transaction amounts
def validate_transaction(amount):
    print(f"Transaction Amount: {amount}")
    if amount > 50000:
        print("Transaction Status: Rejected\n")
    else:
        print("Transaction Status: Approved\n")

# Test the function
validate_transaction(60000)
```

```python
# Problem 9: Calculate attendance and exam eligibility
def check_exam_eligibility(attendance_record):
    present_days = 0
    for status in attendance_record:
        if status == "P":
            present_days += 1

    percentage = (present_days / len(attendance_record)) * 100
    print(f"Attendance Percentage: {percentage}")

    if percentage >= 75:
        print("Exam Eligibility: Eligible\n")
    else:
        print("Exam Eligibility: Not Eligible\n")

# Test the function
check_exam_eligibility(["P", "P", "P", "A", "P"])
```

```python
# Problem 10: Calculate bill based on units and classify usage
def calculate_electricity_bill(units):
    bill = 0

    # Calculate slab rates
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    # Classify usage
    if bill < 500:
        status = "Low Usage"
    elif bill <= 1500:
        status = "Moderate Usage"
    else:
        status = "High Usage"

    print(f"Units Consumed: {units}")
    print(f"Total Bill: ₹{bill}")
    print(f"Usage Classification: {status}\n")

# Test the function
calculate_electricity_bill(250)
```
