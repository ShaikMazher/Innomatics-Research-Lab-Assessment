```python
# Problem 1: Find top performers and handle ties
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

# Find the maximum score
max_score = max(employees.values())

# List comprehension to get all names matching the max score
top_performers = [name for name, score in employees.items() if score == max_score]

# Format and print the output
names_string = ", ".join(top_performers)
print(f"Top Performers Eligible for Bonus: {names_string} (Score: {max_score})")
```

```python
# Problem 2: Count keyword frequency greater than 1
query = "Buy mobile phone buy phone online"

# Convert to lowercase and split into words
words = query.lower().split()
word_counts = {}

# Count the frequency of each word
for w in words:
    if w in word_counts:
        word_counts[w] += 1
    else:
        word_counts[w] = 1

# Filter for keywords searched more than once
duplicates = {w: count for w, count in word_counts.items() if count > 1}
print(duplicates)
```

```python
# Problem 3: Identify valid readings (even numbers) and store with index
sensor_readings = [3, 4, 7, 8, 10, 12, 5]
valid_readings = []

# Loop through the list with enumerate to get index and value
for index, value in enumerate(sensor_readings):
    if value % 2 == 0:  # Check if even
        valid_readings.append((index, value))

print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

```python
# Problem 4: Calculate percentage usage of email domains
emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

domain_counts = {}
total_emails = len(emails)

# Extract domain and count occurrences
for email in emails:
    domain = email.split('@')[1]
    if domain in domain_counts:
        domain_counts[domain] += 1
    else:
        domain_counts[domain] = 1

# Calculate and print percentage
for domain, count in domain_counts.items():
    percentage = (count / total_emails) * 100
    print(f"{domain}: {int(percentage)}%")
```

```python
# Problem 5: Detect days where sales are >30% above average
sales = [1200, 1500, 900, 2200, 1400, 3000]

# Calculate average sales
average_sales = sum(sales) / len(sales)

# Calculate the threshold (30% above average)
threshold = average_sales * 1.30

# Find and print the spikes
for i, sale in enumerate(sales):
    if sale > threshold:
        # i + 1 is used because days start at 1, not 0
        print(f"Day {i + 1}: {sale}")
```

```python
# Problem 6: Identify duplicate user IDs and their counts
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]
id_counts = {}

# Count occurrences of each ID
for uid in user_ids:
    if uid in id_counts:
        id_counts[uid] += 1
    else:
        id_counts[uid] = 1

# Display only the duplicates
for uid, count in id_counts.items():
    if count > 1:
        print(f"{uid} {count} times")
```
